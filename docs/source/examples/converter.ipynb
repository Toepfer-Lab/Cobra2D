# Converter

This Jupyter Notebook introduces the included converters. Their task is to translate Cobra models into other formats, so that the information to create a graph is preserved.

First, we load the textbook model provided by COBRApy, which we convert in the following.

```python
from cobra.io import load_model
from cobra2d.visualization import converter

textbook = load_model("textbook")
```

## IGraph

`cobra2igraph` translates a model into an `igraph.Graph`. Metabolites and reactions both become vertices. Each edge is added from a substrate to the reaction or from the reaction to a product and stores the stoichiometric coefficient (negative for substrates) in the edge attribute `coeff`. Note that the resulting graph is undirected.

```python
graph = converter.cobra2igraph(textbook)
print(graph.summary())
```

## NetworkX

`cobra2networkx` creates the same structure as a directed `networkx.DiGraph`. The node attribute `type` distinguishes metabolites and reactions, and the edge attribute `coeff` contains the absolute value of the stoichiometric coefficient.

```python
graph = converter.cobra2networkx(textbook)
print(graph)
list(graph.in_edges("PFK", data=True))
```

## MetExplore

MetExploreViz reads the model from a JSON file and the flux values from a separate file.

For the representation of the model itself, we can either create the JSON string or directly create a file containing it. In the following, first the string is created as an example and then the file.

```python
JSON_string = converter.cobra2metexplore(textbook)
print(JSON_string[0:373] + "...")
```

```python
import os

converter.cobra2metexplore_file(textbook, "./textbook.json")

os.listdir()
```

Now we create a tab-separated file containing the flux values, which can later be read by MetExploreViz.

We create a solution for the problem and translate it into a file that can be read by MetExploreViz.

```python
solution = textbook.optimize()

converter.cobra2metexplore_flux_file(solution, "./textbook_flux.tsv")

os.listdir()
```
