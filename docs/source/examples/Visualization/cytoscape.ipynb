# Visualization

In the following, the visualization using Cytoscape as well as the interface for Metexplore will be demonstrated with an example.

First, we create a Constraints object that contains both a linker and a phase that has a specific model associated with it.

```python
from model_duplication import Constraints, Linker
from cobra.io import read_sbml_model

con = Constraints()
con.add_sub_models(["root", "leaf", "leaf_2"], [1, 2, 1])
con.add_time_slots(5, 2)
con.add_linker_series("ATP",last2first=True)
con.add_linker_series("ADP",last2first=True)

con.add_linker(
    Linker(
        "NADH",
        source="leaf_2-2",
        destination="leaf_2-3"
    )
)

model = read_sbml_model("/home/jan/Downloads/PlantCoreMetabolism_v2_0_0.xml")
model.solver = 'glpk'
con.get_phase_by_id("leaf-0").model = model
```

Now we can create the visualization using Cytoscape with a command.

```python
con.cytoscape()
```

Now let's take a look at the interface for MetExploreViz. On the left side all groups of the model are shown and can be selected. Selecting a group will cause it to be displayed. On the right side we see all metabolites. If these are selected, they will be hidden in the later display.

```python
con.cytoscape()
```
