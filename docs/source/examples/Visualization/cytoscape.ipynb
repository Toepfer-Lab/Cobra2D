# Visualization using Cytoscape and MetExploreViz

This example demonstrates the Cytoscape visualization and the interface for MetExploreViz.

First, we create a Constraints object that contains a linker reaction and a phase with its own assigned model.

```python
from cobra2d import Constraints, Linker
from cobra.io import load_model

con = Constraints()
con.add_sub_models(["root", "leaf", "stem"], [1, 2, 1])
con.add_time_slots(5, 2)
con.add_linker_series("ATP",last2first=True)
con.add_linker_series("ADP",last2first=True)

con.add_linker(
    Linker(
        "NADH",
        source="stem_2",
        destination="stem_3"
    )
)

model = load_model("textbook")
con.get_phase_by_id("leaf_0").model = model
```

The Cytoscape visualization is created with a single command.

```python
con.cytoscape()
```

Now let's take a look at the interface for MetExploreViz. The left side lists all groups of the model; selected groups will be displayed. The right side lists all metabolites; selected metabolites are treated as side metabolites and hidden in the visualization.

```python
from cobra2d.visualization.helper import metexplore_interface

metexplore_interface(model)
```
