# Visualization using GraphViz

```python
from cobra import Configuration
from cobra2d import Constraints

cobra_config = Configuration()
cobra_config.solver = "glpk"

con = Constraints()
con.add_sub_models(["root", "leaf", "stem"], [1, 2, 1])
con.add_time_slots(5, 2)
```

```python
con.create_graph()
```

```python
from cobra2d import Linker

con.add_linker_series("ATP",last2first=False)
con.add_linker(
    Linker(
        "ADP",
        source="stem_3",
        destination="stem_4"
    )
)
con.create_graph()
```
