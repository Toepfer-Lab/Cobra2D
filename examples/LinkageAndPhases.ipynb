```python
from cobra.test import create_test_model

from model_duplication.constraints.linker import Linkage, Linker
from model_duplication.constraints.phase import Phases, Phase
```

```python
phases = Phases()

for name in ["root", "leaf"]:
    for time in range(1,4):
        phases.phases.append(
            Phase(
                id = name +"@"+str(time),
                light_dark="light",
                timeframe=1,
                volume=1
            )
        )
```

```python
print(phases)
```

```python
model = create_test_model(model_name="textbook")

print(len(model.reactions))
```

```python
model = phases.apply_phases(model)
phases.phases[1].volume = 2

print(len(model.reactions))
```

```python
linkage = Linkage()
linkage.linker.append(
    Linker(
        source=phases.phases[1],
        destination=phases.phases[2],
        id="acald_e"
    )
)

linkage.apply_linkage(model, phases)
print(len(model.reactions))
```
