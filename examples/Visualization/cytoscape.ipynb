```python
from model_duplication import Constraints, Linker
from cobra.io import read_sbml_model

con = Constraints()
con.add_sub_models(["root", "leaf", "leaf_2"], [1, 2, 1])
con.add_time_slots(5, 2)
con.add_linker_series("ATP",last2first=True)
con.add_linker_series("ADP",last2first=True)

con.add_linker(
    Linker(
        "NADH",
        source="leaf_2-2",
        destination="leaf_2-3"
    )
)

model = read_sbml_model("/home/jan/Downloads/PlantCoreMetabolism_v2_0_0.xml")
model.solver = 'glpk'
con.get_phase_by_id("leaf-0").model = model
```
